```python
#Delete all the rows where Director column has missing values.
df.dropna(subset=['director'])
```

```python
#Print all the records where country has Germany value (including West Germany). If any other country is there along with Germany, then that row should also come in output.
df[df['country'].str.contains('Germany', case=False, na=False)]
```

```python
#Expand Duration column into 2 separate columns – First column having the numeric value and other having String. Eg: 3 seasons should be split in 2 columns having 3 in 1st column and seasons in 2nd column.
df[['Duration_Value', 'Duration_Type']] = (df['duration'].str.extract(r'(\d+)\s+(.*)'))
```

```python
df['Duration_Value']=df['Duration_Value'].astype(int)
```

```python
df['Duration_Value'].dtypes
```

```python
# Split Date added into 3 separate columns having date value in 1st column, month value in 2nd column and year value in 3rd.
df[['Month', 'Day', 'Year']] = df['date_added'].str.extract( r'([A-Za-z]+)\s+(\d+),\s+(\d+)')
```

```python
# Print the number of TV shows/Movies released in each year.
df.groupby('release_year')['show_id'].count()
```

```python
#Rename the column title with movie_title
df.rename(columns={'tile':'movie_title'},inplace=True)
```

```python
#Split Listed_in column into 3 different columns with col name (Genre1, Genre2, Genre3). Split the column based on comma.
df[['Genre1', 'Genre2', 'Genre3']] = df['listed_in'].str.split(',', n=2, expand=True)
```

***Import both the attached files (student.csv and marks.csv) in Jupyter notebook and perform following operations:***

```python
# import the files
student_df=pd.read_csv('student.csv')
marks_df=pd.read_csv('marks.csv')
```

```python
#Combine both the dataframes into single dataframe which will have all the records from both the tables.
print(f"The Student dataset \n {student_df.head()}")
print(f"The Marks dataset \n {marks_df.head()}")
```

```python
#checking for null values
student_df.isnull().sum()
```

```python
marks_df.isnull().sum()
```

```python
# Join two tables
combined_df=pd.merge(student_df,marks_df,on='Student_id',how='outer')
combined_df
```

```python
#Print the maximum and minimum marks Gender wise.

combined_df.groupby('Gender')['Mark'].agg(['min','max'])
```

```python
#Print all the students IDs and their marks who have scored more than the average marks of the class.
combined_df[combined_df['Mark']> (combined_df['Mark'].sum())/len(combined_df)]
```

```python
#Print the dataframe who are Males and are Employed.
combined_df[(combined_df['Gender']=="Male") & (combined_df['Employed']=="yes")]
```

```python
#Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak). If student scored than 80 then Tag him as Intelligent, if student scored between 50-80, then Mediocre, else weak.


conditions = [combined_df['Mark'] > 80,combined_df['Mark'].between(50, 80, inclusive='both')]

choices = ['Intelligent', 'Mediocre']

combined_df['IQ_level'] = np.select(conditions,choices,default='Weak')
combined_df['IQ_level']
```

```python
#Count the number of males and females from each city.
combined_df.groupby(['City','Gender'])['Student_id'].count()
```

```python
#Print the top 5 Male scorers.
male_top5 = ( combined_df[combined_df['Gender'] == 'Male'] .sort_values(by='Mark', ascending=False).head(5))
male_top5
```

```python
#Replace the Male value with M and Female value with F and export this dataframe to excel file in D: (D drive) and name the file as test.csv.

combined_df['Gender']=combined_df['Gender'].replace({'Male':'M','Female':'F'})
combined_df
```
